```python
from pylumo import pyLumo, LumoTools, LumoModel, login_via_proton_auth
from dotenv import find_dotenv, load_dotenv
import os

# ## Guest mode (no authentication)
# client = pyLumo()
# response = client.send_request("Hello, Lumo!")
# print(response["message"])

load_dotenv(find_dotenv(usecwd=True))  # Search upward from kernel CWD for the repo-root .env
PROTON_PW = os.environ["PROTON_PW"]  # the actual password string (load_dotenv returns a bool, not the value)
PROTON_USER_NAME = os.environ["PROTON_USER_NAME"]  # the actual username string (load_dotenv returns a bool, not the value)

## Authenticated mode
client = pyLumo()
login_via_proton_auth(client=client, 
                      username=os.environ["PROTON_USER_NAME"],  # can be username or user's email address 
                      password=PROTON_PW, 
                      mailbox_password=os.environ["PROTON_MAILBOX_PW"]  # delegates SRP login + CAPTCHA solve to proton-auth
                      ) 
client.save_session("~/.pylumo_session.json")

# With tools enabled
response = client.send_request(
    "What's the weather in Auckland?",
    tools=[LumoTools.WEATHER, LumoTools.WEB_SEARCH],
    model=LumoModel.LUMO_LITE
)
```
